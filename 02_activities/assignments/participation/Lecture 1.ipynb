```python
# Load packages
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
# Get data to plot
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100, 50)
```

```python
# Draw a scatter plot
fig, ax = plt.subplots(figsize = (5, 3))
ax.scatter(x, y)
```

```python
# Draw a bar plot
fig, ax = plt.subplots(figsize = (5, 3))
ax.bar(x, y)
```

```python
# Draw a line plot
fig, ax = plt.subplots(figsize = (5, 3))
ax.plot(x, y)
```

```python
# Draw a histogram
fig, ax = plt.subplots(figsize = (5, 3))
ax.hist(y)
```

```python
# Example of using axis labels and titles
fig, ax = plt.subplots(figsize = (5, 3))
ax.plot(x, y)
ax.set_title("Total growth over time")
ax.set_ylabel("Total growth")
ax.set_xlabel("Years since start")
fig.tight_layout()
```

```python
# Example of changing fonts for the plot titles
font1 = {"family":"sans-serif", "color":"blue", "size":20}
font2 = {"family":"monospace", "color":"green", "size":14}

fig, ax = plt.subplots(figsize = (5, 3))
ax.plot(x, y)
ax.set_title("Total growth over time", fontdict = font1)
ax.set_ylabel("Total growth", fontdict = font2)
ax.set_xlabel("Years since start", fontdict = font2)
fig.tight_layout()
```

```python
# Example of modifying data points
fig, ax = plt.subplots(figsize = (5, 3))
ax.scatter(x, 
           y, 
           marker = "*",
           color = "indigo")
fig.show()
```

```python
# Connect data points with a line
fig, ax = plt.subplots(figsize = (5, 3))
ax.plot(x, 
           y, 
           marker = "*",
           color = "indigo",
           linestyle = "--", 
           linewidth = 2)
fig.show()
```

```python
# Using hex codes for colors
fig, ax = plt.subplots(figsize = (5, 3))
ax.plot(x, 
           y, 
           marker = "*",
           color = "#7425b9",
           linestyle = "--", 
           linewidth = 2)
fig.show()
```

```python
# Further modifying data points
fig, ax = plt.subplots(figsize = (5, 3))
ax.plot(x, 
           y, 
           marker = "*",
           color = "#7425b9",
           linestyle = "--", 
           linewidth = 2, 
           markersize = 12,
           markeredgecolor = "#fa9359", 
           markerfacecolor = "#000000")
fig.show()
```

```python
# Add in grid lines
fig, ax = plt.subplots(figsize = (5, 3))
ax.plot(x, 
           y, 
           marker = "*",
           color = "#7425b9",
           linestyle = "--", 
           linewidth = 2, 
           markersize = 12,
           markeredgecolor = "#fa9359", 
           markerfacecolor = "#000000")

ax.grid(axis = "y", color = "blue", linewidth = 2, linestyle = "--")

fig.show()
```
