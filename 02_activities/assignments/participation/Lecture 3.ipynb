```python
# Load packages
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
# Generate some data to plot
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0, 100, 50)
y2 = np.random.randint(0, 100, 50)
```

```python
# Plot both datasets as lineplots on the same plot
fig, ax = plt.subplots(figsize = (5, 3))
ax.plot(x, y1)
ax.plot(x, y2)
fig.show()
```

```python
# Add a legend to the lower-right of the plot
fig, ax = plt.subplots(figsize = (5, 3))
ax.plot(x, y1, label = "Person 1")
ax.plot(x, y2, label = "Person 2")
ax.legend(loc = "lower right")
fig.show()
```

```python
# Modify our legend
fig, ax = plt.subplots(figsize = (5, 3))
ax.plot(x, y1, label = "Person 1")
ax.plot(x, y2, label = "Person 2")
ax.legend(loc = "lower right",
          frameon = True, # add a frame
          fontsize = 12, 
          ncol = 2, # 2 columns
          shadow = True) # shadow around the frame
fig.show()
```

```python
# Move legend outside of the plot area
fig, ax = plt.subplots(figsize = (5, 3))
ax.plot(x, y1, label = "Person 1")
ax.plot(x, y2, label = "Person 2")
ax.legend(loc = "upper left", 
          bbox_to_anchor = (1, 1)) # 1, 1 = top-right of plot
fig.show()
```

```python
# Add text annotations with text()
fig, ax = plt.subplots(figsize = (5, 3))
ax.scatter(x, y1, label = "Person 1")
ax.scatter(x, y2, label = "Person 2")
ax.legend(loc = "lower right")

ax.text(10, 95, "This value is important!") #10 on x-axis, 95 on y-axis
fig.show()
```

```python
# Modify our text annotations
fig, ax = plt.subplots(figsize = (5, 3))
ax.scatter(x, y1, label = "Person 1")
ax.scatter(x, y2, label = "Person 2")
ax.legend(loc = "lower right")

ax.text(10, 95, "This value is important!",
        ha = "center", # Horizontal alignment of text
        color = "red", 
        size = 20) 
fig.show()
```

```python
# Positioning text annotations relative to data/axes/figure
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])

ax.text(1, 5, ". Data:(1, 5)", transform = ax.transData)
ax.text(0.5, 0.1, ". Axes:(0.5, 0.1)", transform = ax.transAxes)
ax.text(0.5, 0.1, ". Figure:(0.5, 0.1)", transform = fig.transFigure)
```

```python
# Annotating with arrows
fig, ax = plt.subplots(figsize = (5, 3))
ax.scatter(x, y1, label = "Person 1")
ax.scatter(x, y2, label = "Person 2")
ax.legend(loc = "lower right")

ax.annotate("This is important", 
            xy = (10, 95), # Arrow position
            xytext = (20, 94), # Text position
            arrowprops = dict(facecolor = "black")) # Define the arrow
fig.show()
```

```python
# Modifying arrows
fig, ax = plt.subplots(figsize = (5, 3))
ax.scatter(x, y1, label = "Person 1")
ax.scatter(x, y2, label = "Person 2")
ax.legend(loc = "lower right")

ax.annotate("This is important", 
            xy = (10, 95), # Arrow position
            xytext = (20, 94), # Text position
            arrowprops = dict(arrowstyle = "wedge", facecolor = "hotpink")) # Define the arrow
fig.show()
```

```python
# Removing tick marks and labels
fig, ax = plt.subplots(figsize = (5, 3))
ax.scatter(x, y1, label = "Person 1")
ax.scatter(x, y2, label = "Person 2")
ax.legend(loc = "lower right")

ax.yaxis.set_major_locator(plt.NullLocator()) # No y-axis ticks or labels
ax.xaxis.set_major_formatter(plt.NullFormatter()) # No x-axis labels, but still ticks

fig.show()
```

```python
# Controlling the number of tick marks
fig, ax = plt.subplots(figsize = (5, 3))
ax.scatter(x, y1, label = "Person 1")
ax.scatter(x, y2, label = "Person 2")
ax.legend(loc = "lower right")

ax.xaxis.set_major_locator(plt.MaxNLocator(3)) # Only 3 ticks

fig.show()
```

```python
# Set tick mark intervals
fig, ax = plt.subplots(figsize = (5, 3))
ax.scatter(x, y1, label = "Person 1")
ax.scatter(x, y2, label = "Person 2")
ax.legend(loc = "lower right")

ax.xaxis.set_major_locator(plt.MultipleLocator(5)) # Ticks every 5 units

fig.show()
```

```python
# Rotate axis labels
fig, ax = plt.subplots(figsize = (5, 3))
ax.scatter(x, y1, label = "Person 1")
ax.scatter(x, y2, label = "Person 2")
ax.legend(loc = "lower right")

plt.xticks(rotation = 45, ha = "right") # Rotate x-axis ticks by 45 degrees

fig.show()
```
