```python
# Load packages
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
# Generate some data to plot
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0, 100, 50)
y2 = np.random.randint(0, 100, 50)
```

```python
# Change axis title fonts
fig, ax = plt.subplots(figsize = (5, 3))
ax.scatter(x, y1, label = "Person 1")
ax.scatter(x, y2, label = "Person 2")
ax.legend(loc = "lower right")

font1 = {"family":"serif", "color":"indigo"}
# ax.set_xlabel("New X axis!", fontsize = 18, fontdict = font1)
plt.xlabel("New X axis!", fontsize = 18, fontdict = font1)
fig.show()
```

```python
# Get some new data to plot
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75, 50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
# Make a plot with multiple subplots
fig, (ax1, ax2) = plt.subplots(ncols = 2, 
                               nrows = 1, 
                               figsize = (7, 3))

ax1.scatter(x1, y1)
ax2.bar(x2, y2)

fig.show()
```

```python
# Annotate one of the subplots
fig, (ax1, ax2) = plt.subplots(ncols = 2, 
                               nrows = 1, 
                               figsize = (7, 3))

ax1.scatter(x1, y1)
ax1.annotate("This is important", xy = (10, 65), 
             xytext = (30, 65), 
             arrowprops = dict(facecolor = "black") )
ax2.bar(x2, y2)

fig.show()
```

```python
# Plot 3 subplots on one figure
fig, someaxes = plt.subplot_mosaic([["ax1", "ax3"], # First row - axes 1 and 3
                                    ["ax2", "ax3"]], # Second row - axes 2 and 3
                                    figsize = (7, 4))

someaxes["ax1"].scatter(x1, y1)
someaxes["ax2"].bar(x2, y2)
someaxes["ax3"].plot(x1, y1)

plt.show()
```

```python
# Set axis labels for subplots
fig, someaxes = plt.subplot_mosaic([["ax1", "ax3"], 
                                    ["ax2", "ax3"]], 
                                    figsize = (7, 4))

someaxes["ax1"].scatter(x1, y1)
someaxes["ax2"].bar(x2, y2)
someaxes["ax3"].plot(x1, y1)

someaxes["ax1"].set_xlabel("A Big Label", fontsize = 18) # Doesn't actually appear
someaxes["ax2"].set_xlabel("Another Label", fontsize = 18)
someaxes["ax3"].set_xlabel("Label 2: 2 Fast 2 Furious", fontsize = 18) # Gets cut off
```

```python
# Use layouts to organize subplots
fig, someaxes = plt.subplot_mosaic([["ax1", "ax3"], 
                                    ["ax2", "ax3"]], 
                                    figsize = (7, 4), 
                                    layout = "constrained")

someaxes["ax1"].scatter(x1, y1)
someaxes["ax2"].bar(x2, y2)
someaxes["ax3"].plot(x1, y1)

someaxes["ax1"].set_xlabel("A Big Label", fontsize = 18)
someaxes["ax2"].set_xlabel("Another Label", fontsize = 18)
someaxes["ax3"].set_xlabel("Label 2: 2 Fast 2 Furious", fontsize = 18)

fig.show()
```

```python
# Put multiple subplots on one axes object

# Make sample data
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

# Make one figure with one axis
fig, ax = plt.subplots(figsize = (7, 3))

# Make both plots on the same axes
ax.bar(x, y1, color = "indigo")
ax.plot(x, y2, color = "red")

fig.show()
```

```python
# Add in error information
y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize = (7, 3))

# Add the actual data
ax.plot(x, y2, color = "red")

# Add the error bars
ax.errorbar(x, 
            y2, 
            yerr = y2_sd, 
            color = "red",
            fmt = "none")

fig.show()
```

```python
# Customizing error bars
fig, ax = plt.subplots(figsize = (7, 3))

# Add the actual data
ax.plot(x, y2, color = "red")

# Add the error bars
ax.errorbar(x, 
            y2, 
            yerr = y2_sd, 
            color = "indigo",
            fmt = "none", 
            elinewidth = 4, 
            capsize = 6, 
            capthick = 4)

fig.show()
```

```python
# Only show error bars in intervals
fig, ax = plt.subplots(figsize = (7, 3))

# Add the actual data
ax.plot(x, y2, color = "red")

# Add the error bars
ax.errorbar(x, 
            y2, 
            yerr = y2_sd, 
            color = "indigo",
            fmt = "none", 
            elinewidth = 4, 
            capsize = 6, 
            capthick = 4,
            errorevery = 2) # Show every other error bar

fig.show()
```

```python
# Load some libraries
from PIL import Image
import requests
from io import BytesIO
```

```python
# Use images from the Internet
url = "https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png"

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "image/*,*/*;q=0.8",
    "Referer": "https://en.wikipedia.org/"
}

response = requests.get(url, headers = headers)
image_file = BytesIO(response.content)
image = Image.open(image_file)
image
```

```python
# Add the image to a plot
fig, ax = plt.subplots(figsize = (7, 3))

ax.plot(x, y2, color = "red")

ax_image = fig.add_axes([0.1, # x coordinate
                         0.11, # y coordinate
                         0.15, # image width
                         0.35]) # image height

ax_image.imshow(image)
ax_image.axis("off") # turn off axes for the image

fig.show()
```
