```python
# Load packages
import matplotlib.pyplot as plt
import matplotlib.dates as dates
import pandas as pd

# matplotlib version: 3.10.6
# pandas version: 2.3.2

# Load in the dataset to be plotted
fuel_data = pd.read_csv("./fueltypesall.csv")

# Get just the rows corresponding to the 5 fuel types I want to plot (diesel, auto propane, regular gasoline, premium gasoline, and natural gas)
fuel_data_diesel = fuel_data.loc[fuel_data["Fuel Type"] == "Diesel"]
fuel_data_propane = fuel_data.loc[fuel_data["Fuel Type"] == "Auto Propane"] 
fuel_data_regular_gas = fuel_data.loc[fuel_data["Fuel Type"] == "Regular Unleaded Gasoline"] 
fuel_data_premium_gas = fuel_data.loc[fuel_data["Fuel Type"] == "Premium Gasoline"] 
fuel_data_natural_gas = fuel_data.loc[fuel_data["Fuel Type"] == "Compressed Natural Gas"] 

# Make a line plot of prices over time
fig, ax = plt.subplots(figsize = (10, 6))

# Plot the 5 different fuel types as separate lines, converting the x-axis values to dates
x_premium_gas = pd.to_datetime(fuel_data_premium_gas["Date"])
y_premium_gas = fuel_data_premium_gas["Ontario Average/Moyenne provinciale"] / 100
ax.plot(x_premium_gas, y_premium_gas, label = "Premium Gasoline", alpha = 0.75)

x_regular_gas = pd.to_datetime(fuel_data_regular_gas["Date"])
y_regular_gas = fuel_data_regular_gas["Ontario Average/Moyenne provinciale"] / 100
ax.plot(x_regular_gas, y_regular_gas, label = "Regular Unleaded Gasoline", alpha = 0.75)

x_diesel = pd.to_datetime(fuel_data_diesel["Date"])
y_diesel = fuel_data_diesel["Ontario Average/Moyenne provinciale"] / 100
ax.plot(x_diesel, y_diesel, label = "Diesel", alpha = 0.75, color = "purple")

x_natural_gas = pd.to_datetime(fuel_data_natural_gas["Date"])
y_natural_gas = fuel_data_natural_gas["Ontario Average/Moyenne provinciale"]/ 100
ax.plot(x_natural_gas, y_natural_gas, label = "Compressed Natural Gas", alpha = 0.75, color = "black")

x_propane = pd.to_datetime(fuel_data_propane["Date"])
y_propane = fuel_data_propane["Ontario Average/Moyenne provinciale"] / 100
ax.plot(x_propane, y_propane, label = "Auto Propane", alpha = 0.75)


# Format the intervals for the axes
ax.xaxis.set_major_locator(dates.YearLocator(2))
ax.xaxis.set_major_formatter(dates.DateFormatter("%Y"))
ax.yaxis.set_major_locator(plt.MultipleLocator(0.25))

# Set the legend
ax.legend(loc = "upper left", fontsize = 14, framealpha = 1)

# Format axis tick labels
plt.xticks(rotation = 45, ha = "right", fontsize = 12)
plt.yticks(fontsize = 12)

# Set the titles
plt.xlabel("Time", fontsize = 18)
plt.ylabel("Fuel price (dollars per litre)", fontsize = 18)
plt.title("Average Ontario fuel prices over time", fontsize = 20)

# Turn on gridlines
plt.grid(linewidth = 0.5)

# Show the finalized plot
plt.show()
```
